```python
# print Odd Numbers in the list
num= (20,10,16,19,25,1,276,188)
print(num)
for i in num:
    if(i%2==1):
        print(i,"is odd")
```

```python
# print Even numbers in the list
num= (20,10,16,19,25,1,276,188)
print(num)
for i in num:
    if(i%2==0):
        print(i,"is even")
```
