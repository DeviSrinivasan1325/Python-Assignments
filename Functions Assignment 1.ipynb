```python
# Create a function that checks whether the given number is Odd or Even
class OddEvenCheck:
    def OddEven(self):
        num= int(input("Enter a number:"))
        if(num%2==0):
            print(num,"is Even number")
        else:
            print(num,"is Odd number")
OddEven= OddEvenCheck()
```

```python
OddEven.OddEven()
```

```python
# Create a function that tells elegibility of marriage for male and female according to their age limit like 21 for male and 18 for female
class MarriageElegibilty:
    def Elegible(self):
        Gender= input("Your Gender:")
        Age= int(input("Your Age:"))
        if (Gender=="Male" and Age>=21):
            print("Elegible")
        elif(Gender=="Female" and Age>=18):
            print("Elegible")
        else:
            print("Not Elegible")
ElegiblityForMarriage= MarriageElegibilty()
```

```python
ElegiblityForMarriage.Elegible()
```

```python
# calculate the percentage of your 10th mark
class markpercentage:
    def percentage(self):
        sub1= int(input("Subject1="))
        sub2= int(input("Subject2="))
        sub3= int(input("Subject3="))
        sub4= int(input("Subject4="))
        sub5= int(input("Subject5="))
        totalmarks= sub1+sub2+sub3+sub4+sub5
        print("Total:",totalmarks)
        per= (totalmarks / 500)*100
        print("Percentage:",per)
FindPercent= markpercentage()
```

```python
FindPercent.percentage()
```

```python
#print area and perimeter of triangle using class and functions
class ShapeMeasurement:
    def triangle(self):
        height= int(input("Height:"))
        base= int(input("Base:"))
        print("Area formula: (Height*Base)/2")
        Area= (height*base)/2
        print("Area of triangle:",Area)
        side1=int(input("Side1:"))
        side2=int(input("Side2:"))
        base=int(input("Base:"))
        print("Perimeter formula: Side1+Side2+Base")
        Perimeter= side1+side2+base
        print("Perimeter of Triangle:",Perimeter)
triangle= ShapeMeasurement()        
```
